# Simple interactive chat example

```python
from utils.multiturn_chat import add_user_message, add_assistant_message, chat
```

```python
# initialize list of messages
messages = []

# start the while loop until an interruption takes place
while True:
    # get user input
    user_input = input("> ")
    print(">", user_input)
    print('-'*10)
    # 
    add_user_message(messages=messages, text=user_input)
    #
    answer = chat(messages_lst=messages)
    print(answer.replace("**"," "))
    print('-'*10)
    #
    add_assistant_message(messages=messages, text=answer)

```
