Create an API client and make a request. Make sure to check the models available in Claude API console

```python
# Create an API client
from anthropic import Anthropic
```

```python
client = Anthropic()
model_selected = "claude-sonnet-4-6"
```

Creating auxiliary functions

```python
def add_user_message(messages: list, text: str) -> None:
    user_message = {"role": "user", "content": text}
    messages.append(user_message)
```

```python
def add_assistant_message(messages: list, text: str) -> None:
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
def chat(messages_lst: list[str]) -> str:
    # Make a request
    message = client.messages.create(
        model=model_selected, # model selected by user, setup in cell above
        max_tokens=1000, # max budget on number of tokens that Claude can generate. 
        messages=messages_lst
        )
    return message.content[0].text
```

### Showing capabilities of functions above

```python
# make a starting list of messages
messages = []
```

```python
# Add in the initial user question: "Define quantum computing in one sentence"
add_user_message(messages, 'Define quantum computing in one sentence')  
print(messages)
```

```python
# Pass the list of messages into the chat to get an answer
answer_01 = chat(messages)
print(answer_01)
```

```python
# take the answer and add it as an assistant message into our list
add_assistant_message(messages, answer_01)
```

```python
messages
```

```python
# Add in the user's follow-up question
add_user_message(messages, "Write another sentence")
```

```python
# Call chat again with the list of messages to get a final answer
answer_02 = chat(messages)
print(answer_02)
```
